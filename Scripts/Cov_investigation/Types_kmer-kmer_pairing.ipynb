```python
from Helper_functions import Build_canonical_map
Canonical_map = Build_canonical_map(4)
```

```python
import pandas as pd
import re 

# Create a two by two matrix 

def kmer_overlap_pair(kmer1, kmer2):
    k = len(kmer1)
    
    # Leading overlap (prefix of kmer1 vs suffix of kmer2)
    leading = 0
    for n in reversed(range(1, k+1)):
        if kmer1[:n] == kmer2[-n:]:
            leading = n
            break
    
    # Trailing overlap (suffix of kmer1 vs prefix of kmer2)
    trailing = 0
    for n in reversed(range(1, k+1)):
        if kmer1[-n:] == kmer2[:n]:
            trailing = n
            break
    
    return (leading, trailing)
    
# List of canonical kmers
Canonical_kmers = sorted(list(set(Canonical_map.values())))

# Initialize empty dataframe
df = pd.DataFrame(index=Canonical_kmers, columns=Canonical_kmers)

# Fill in the matrix
for kmer1 in Canonical_kmers:
    for kmer2 in Canonical_kmers:
        L, T = kmer_overlap_pair(kmer1, kmer2)
        df.loc[kmer1, kmer2] = (L, T) 
```
